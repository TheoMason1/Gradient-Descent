```python
#This program is a Gradient Descent program which finds the minima of a function by implemetnting gradient descent 
#at a number of random starting points within a (sensible) range chosen by the user.
#In this case we choose f(x,y) = sin(x+y) + (x-y)^2 - 1.5x + 2.5y + 1
#I have been using Jupyter notebook but it should work elsewhere.

import numpy as np
import matplotlib.pyplot as plt



def f(x,y):
    return (np.sin(x+y)) + (x-y)**2 - 1.5*x + 2.5*y + 1               #define function, in this case f(x,y) = sin(x+y) + (x-y)^2 - 1.5x + 2.5y + 1


X = np.linspace(-1,4,num=40, endpoint=True)                          #use x,y for inputs to function and X,Y for the domain
Y = np.linspace(-3,3,num=40, endpoint=True)                                            

XX, YY = np.meshgrid(X, Y)        
A = f(XX,YY)                                                         #value of the function at each point

plt.figure(1)
plt.plot(X, f(X,Y))                                                   #plot f(x)
plt.xlabel('$X$')
plt.ylabel('$f(x,y)$')
plt.title('$f(x,y)$')





def gradientf(x,y):
    return np.array([(np.cos(x+y) + 2*(x-y) - 1.5), (np.cos(x+y) - 2*(x-y) + 2.5)])        #return an array of (df/dx,df/dy)


max_i = 300
precision = 0.00001
N=100


def descent(xn,yn):                                                   #make the gradient descent a function so we can call it repeatably
    step = 0.1
    i = 0
    xpath = []                       
    ypath = []
    c = []
    while step > precision and i < max_i:
             
        rate = 0.01
              
        xnp = xn - rate*(gradientf(xn,yn)[0])
        ynp = yn - rate*(gradientf(xn,yn)[1])
            
            
        xpath.append(xn)
        ypath.append(yn)
                
        step = np.sqrt((xn-xnp)**2+(yn-ynp)**2)                        #pythag to find how far was travelled
        c.append(f(xn,yn))
                            
        xn = xnp                                                       #set current x to new value
        yn = ynp
        i += 1
    return xpath,ypath,c


rng = np.random.default_rng()                                           #set up random starting points and empty lists to store paths
ax = []  
endx = []
ay = []  
endy = []

for n in range(N):                                                      #From random starting points call the gradient descent function N times

    xn = rng.uniform(-1,4)
    yn = rng.uniform(-3,3)

    plt.figure(2)
    plt.scatter(descent(xn,yn)[0], descent(xn,yn)[1], c=descent(xn,yn)[2])       #plot grad descent
    
    ax = (descent(xn,yn)[0])                                                     #save end values so we can find minimum
    endx.append(ax[-1])
    ay = (descent(xn,yn)[1])
    endy.append(ay[-1])

plt.figure(2)                                                         #plot a contour map
plt.contour(XX,YY,A, levels=20)
plt.xlabel('$X$')
plt.ylabel('$Y$')

p = 0
oy = []
ox = []

for p in range(len(endy)):                                              #list that takes values that converge to the global minimum
    if endy[p] < -0.5:
        oy.append(endy[p])
        ox.append(endx[p])
        p+=1

    else:
        pass


oymin = min(oy)                                                       #global minimum
oxmin = min(ox)
print('The global minimum occurs at' , oxmin, oymin)


plt.figure(3)
plt.scatter(endx, endy, marker='.')
plt.title('Endpoints of each gradient descent trial')
plt.xlabel('$X$')
plt.ylabel('$Y$')

plt.figure(4)
plt.scatter(ox, oy, marker='.')
plt.title('Endpoints of each gradient descent trial that converge towards the global minimum')
plt.xlabel('$X$')
plt.ylabel('$Y$')
plt.show()
```
